```python
# filling missing values in object column by mode(most_frequent Value)

for col in df.select_dtypes(include='object').columns:
    print(col,'\t', df[col].isnull().sum(),'\tReplaced By-> ',df[col].mode()[0])
    df[col]=df[col].fillna(df[col].mode()[0])
```

```python
df.isnull().sum()
```

```python
# filling missing values in float column by mean(Average)

for col in df.select_dtypes(include='float').columns:
    if df[col].isnull().sum()>0:
        print(col,'\t', df[col].isnull().sum(),'\tReplaced By->',round(df[col].mean(),2))
        df[col]=df[col].fillna(round(df[col].mean(),2))
    
```
