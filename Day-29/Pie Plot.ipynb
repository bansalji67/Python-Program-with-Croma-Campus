# PIE PLOT

```python
import matplotlib.pyplot as plt
```

```python
# Syntax
# x=[]
# plt.pie(x)
# plt.show()

```

```python
# Create Pie Chart
x=[10,20,30,40]
y=["c","c++","java","Python"]
plt.pie(x)
plt.show()
```

```python
# add labels
x=[10,20,30,40]
y=["c","c++","java","Python"]
plt.pie(x,labels=y)
plt.show()
```

```python
# how to seperate c part from the graph
x=[10,20,30,40]
y=["c","c++","java","Python"]
ex=[0.3,0.0,0.0,0.0]
plt.pie(x,labels=y,explode=ex)
plt.show()
```

```python
# Change the color
x=[10,20,30,40]
y=["c","c++","java","Python"]
ex=[0.3,0.0,0.0,0.0]
c=["r","b","g","y"]
plt.pie(x,labels=y,explode=ex,colors=c)
plt.show()
```

```python
# show %
x=[10,20,30,40]
y=["c","c++","java","Python"]
ex=[0.3,0.0,0.0,0.0]
c=["r","b","g","y"]
plt.pie(x,labels=y,explode=ex,autopct="%0.1f%%",shadow=True)
plt.show()
```

```python
# how to change the graph size
x=[10,20,30,40]
y=["c","c++","java","Python"]
ex=[0.3,0.0,0.0,0.0]
c=["r","b","g","y"]
plt.pie(x,labels=y,explode=ex,autopct="%0.1f%%",shadow=True,radius=0.8)
plt.show()
```

```python
# change label  distance
x=[10,20,30,40]
y=["c","c++","java","Python"]
ex=[0.3,0.0,0.0,0.0]
c=["r","b","g","y"]
plt.pie(x,labels=y,explode=ex,autopct="%0.1f%%",shadow=True,radius=0.8,labeldistance=1.2)
plt.show()
```

```python
# how to rotate the graph
x=[10,20,30,40]
y=["c","c++","java","Python"]
ex=[0.3,0.0,0.0,0.0]
c=["r","b","g","y"]
plt.pie(x,labels=y,explode=ex,autopct="%0.1f%%",shadow=True,radius=0.8,labeldistance=1.2,startangle=90)
plt.show()
```

```python
# Change the text size
x=[10,20,30,40]
y=["c","c++","java","Python"]
ex=[0.3,0.0,0.0,0.0]
c=["r","b","g","y"]
plt.pie(x,labels=y,explode=ex,autopct="%0.1f%%",shadow=True,radius=1.3,labeldistance=1.2,startangle=90
       ,textprops={"fontsize":15})
plt.show()
```

```python
# Change the digram angle position
x=[10,20,30,40]
y=["c","c++","java","Python"]
ex=[0.3,0.0,0.0,0.0]
c=["r","b","g","y"]
plt.pie(x,labels=y,explode=ex,autopct="%0.1f%%",shadow=True,radius=1.3,labeldistance=1.2,startangle=90
       ,textprops={"fontsize":15},counterclock=False)
plt.show()
```

```python
# CHANGE LINE WIDTH
x=[10,20,30,40]
y=["c","c++","java","Python"]
ex=[0.3,0.0,0.0,0.0]
c=["r","b","g","y"]
plt.pie(x,labels=y,explode=ex,autopct="%0.1f%%",shadow=True,radius=1.3,labeldistance=1.2,startangle=90
       ,textprops={"fontsize":15},counterclock=False,wedgeprops={"linewidth":5,"edgecolor":"m"})
plt.show()
```

```python
# how to rotate the labels
x=[10,20,30,40]
y=["c","c++","java","Python"]
ex=[0.3,0.0,0.0,0.0]
c=["r","b","g","y"]
plt.pie(x,labels=y,explode=ex,autopct="%0.1f%%",shadow=True,radius=1,labeldistance=1.2,startangle=90
       ,textprops={"fontsize":15},counterclock=False,wedgeprops={"linewidth":5},rotatelabels=False)
plt.title("Languages")
plt.show()
```

```python
# add legends
x=[10,20,30,40]
y=["c","c++","java","Python"]
ex=[0.3,0.0,0.0,0.0]
c=["r","b","g","y"]
plt.pie(x,labels=y,explode=ex,autopct="%0.1f%%",shadow=True,radius=1,labeldistance=1.2,startangle=90
       ,textprops={"fontsize":15},counterclock=False,wedgeprops={"linewidth":5},rotatelabels=False)
plt.title("Languages")
plt.legend(loc=3)
plt.show()
```
