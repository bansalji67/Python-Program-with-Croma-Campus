```python
no=[44,36,56,11,34,38,30,23,20,53,23,22,58,30,18,13,57,49, 17, 56, 47, 53, 28, 59,
 24,28,44, 36, 29, 45, 32, 37, 54, 52, 55, 44, 47, 33, 29, 56, 30, 16, 43, 26, 51, 56, 18, 22,
 54, 53]

plt.hist(no,color="g",edgecolor="r")
plt.title("Histogram")
plt.xlabel("Python")
plt.ylabel("No")
plt.show()
```

```python
# how to set the bars

l=[10,20,30,40,50,60]
plt.hist(no,color="b",bins=l,edgecolor="r")

plt.title("Histogram")
plt.xlabel("Python")
plt.ylabel("No")
plt.show()
```

```python
# how to set range parameter
l=[10,20,30,40,50,60]
plt.hist(no,"auto",(0,100),edgecolor="r")

plt.title("Histogram")
plt.xlabel("Python")
plt.ylabel("No")
plt.show()
```

```python
l=[10,20,30,40,50,60]
plt.hist(no,color="b",edgecolor="r")

plt.title("Histogram")
plt.xlabel("Python")
plt.ylabel("No")
plt.show()
```

```python
# how to change the frequncy parameter in reverse order
l=[10,20,30,40,50,60]
plt.hist(no,color="b",edgecolor="r",cumulative=-1)

plt.title("Histogram")
plt.xlabel("Python")
plt.ylabel("No")
plt.show()
```

```python
# how to change the y's bottom parameter
l=[10,20,30,40,50,60]
plt.hist(no,color="b",edgecolor="r",bottom=10)

plt.title("Histogram")
plt.xlabel("Python")
plt.ylabel("No")
plt.show()
```

```python
# how to change the alignment
l=[10,20,30,40,50,60]
plt.hist(no,color="b",bins=l,edgecolor="r",align="mid")

plt.title("Histogram")
plt.xlabel("Python")
plt.ylabel("No")
plt.show()
```

```python
# how to change histogram type or style
l=[10,20,30,40,50,60]
plt.hist(no,color="b",bins=l,edgecolor="r",histtype="stepfilled")

plt.title("Histogram")
plt.xlabel("Python")
plt.ylabel("No")
plt.show()
```

```python
# show barstacked
l=[10,20,30,40,50,60]
plt.hist(no,color="b",bins=l,edgecolor="r",histtype="barstacked")

plt.title("Histogram")
plt.xlabel("Python")
plt.ylabel("No")
plt.show()
```

```python
# how to change the graph in horizentally

l=[10,20,30,40,50,60]
plt.hist(no,color="b",bins=l,edgecolor="r",orientation="horizontal")

plt.title("Histogram")
plt.xlabel("Python")
plt.ylabel("No")
plt.show()
```

```python
# how to reduce the width of bar graph
l=[10,20,30,40,50,60]
plt.hist(no,color="b",bins=l,edgecolor="r",rwidth=0.8)

plt.title("Histogram")
plt.xlabel("Python")
plt.ylabel("No")
plt.show()
```

```python
# how to add labels
l=[10,20,30,40,50,60]
plt.hist(no,color="b",bins=l,edgecolor="r",label="python")

plt.title("Histogram")
plt.xlabel("Python")
plt.ylabel("No")
plt.legend()
plt.show()
```
