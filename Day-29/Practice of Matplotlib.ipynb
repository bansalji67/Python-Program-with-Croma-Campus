# Bar Plot


```python
import matplotlib.pyplot as plt
x=['Python','C','C++','Java']
y=[85,70,60,82]
plt.xlabel('language',fontsize=15)
plt.ylabel('No',fontsize=15)
plt.title('Course Details',fontsize=15)
plt.bar(x,y)
plt.show()
```

```python
# Change of Width
import matplotlib.pyplot as plt
x=['Python','C','C++','Java']
y=[85,70,60,82]
plt.xlabel('language',fontsize=15)
plt.ylabel('No',fontsize=15)
plt.title('Course Details',fontsize=15)
plt.bar(x,y,width=0.4)                    # Change of Width size
plt.show()
```

```python
# How to change the color
import matplotlib.pyplot as plt
x=['Python','C','C++','Java']
y=[85,70,60,82]
plt.xlabel('language',fontsize=15)
plt.ylabel('No',fontsize=15)
plt.title('Course Details',fontsize=15)
plt.bar(x,y,width=0.4,color='y')          # Change the Color of bars
plt.show()
```

```python
# How to change colors for each an every bar
import matplotlib.pyplot as plt
x=['Python','C','C++','Java']
y=[85,70,60,82]
plt.xlabel('language',fontsize=15)
plt.ylabel('No',fontsize=15)
plt.title('Course Details',fontsize=15)
c = ["y","b","m","g"]                         # create variable for store bar colors
plt.bar(x,y,width=0.4,color=c)                #  change bar colors
plt.show()
```

```python
# How to change the x-axis label on Edge
import matplotlib.pyplot as plt
x=['Python','C','C++','Java']
y=[85,70,60,82]
plt.xlabel('language',fontsize=15)
plt.ylabel('No',fontsize=15)
plt.title('Course Details',fontsize=15)
c = ["y","b","m","g"]
plt.bar(x,y,width=0.4,color=c,align="edge")
plt.show()
```

```python
# How to change the color of bar's border with Line Width
import matplotlib.pyplot as plt
x=['Python','C','C++','Java']
y=[85,70,60,82]
plt.xlabel('language',fontsize=15)
plt.ylabel('No',fontsize=15)
plt.title('Course Details',fontsize=15)
c = ["y","b","m","g"]
plt.bar(x,y,width=0.4,color=c,edgecolor='r',linewidth=3)
plt.show()
```

```python
# how to change dotted line of bar graphs border
import matplotlib.pyplot as plt
x=['Python','C','C++','Java']
y=[85,70,60,82]
plt.xlabel('language',fontsize=15)
plt.ylabel('No',fontsize=15)
plt.title('Course Details',fontsize=15)
c = ["y","b","m","g"]
plt.bar(x,y,width=0.4,color=c,edgecolor='r',linewidth=3,linestyle=":")
plt.show()
```

```python
# How to change the graph color slightly light
import matplotlib.pyplot as plt
x=['Python','C','C++','Java']
y=[85,70,60,82]
plt.xlabel('language',fontsize=15)
plt.ylabel('No',fontsize=15)
plt.title('Course Details',fontsize=15)
c = ["y","b","m","g"]
plt.bar(x,y,width=0.4,color=c,edgecolor='r',linewidth=3,linestyle=":",alpha=0.4)
plt.show()
```

```python
# HOW To show label parameter
import matplotlib.pyplot as plt
x=['Python','C','C++','Java']
y=[85,70,60,82]
z=[20,30,40,50]
plt.xlabel('language',fontsize=15)
plt.ylabel('No',fontsize=15)
width=0.2
plt.title('Course Details',fontsize=15)
plt.bar(x,y,width,color='r',label="Course Demands")
plt.bar(x,z,width,color='y',label="Course Demands1")
plt.legend()
plt.show()
```

```python
# How to change the bar graph pattern
import matplotlib.pyplot as plt
import numpy as np


x=['Python','C','C++','Java']
y=[85,70,60,82]
z=[20,30,40,50]

width=0.2
p=np.arange(len(x))
p1=[j+width for j in p]

plt.xlabel('language',fontsize=15)
plt.ylabel('No',fontsize=15)
plt.title('Course Details',fontsize=15)

plt.bar(p,y,width,color='r',label="Course Demands")
plt.bar(p1,z,width,color='y',label="Course Demands1")
plt.xticks(p+width/2,x,rotation=20)
plt.legend()
plt.show()
```

```python
# how to create Horizental bar graph
import matplotlib.pyplot as plt
x=['Python','C','C++','Java']
y=[85,70,60,82]
z=[20,30,40,50]

plt.xlabel('language',fontsize=15)
plt.ylabel('No',fontsize=15)


plt.barh(x,y,color='r',label="Course Demands")  # add h for bar graph
plt.barh(x,z,color='y',label="Course Demands1")
plt.legend()
plt.show()
```
