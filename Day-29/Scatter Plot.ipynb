# Scatter Plot


```python
# Create Scatter Plot
import matplotlib.pyplot as plt
Day=[1,2,3,4,5,6,7]
No=[2,3,1,4,5,3,6]
plt.scatter(Day,No)
plt.show()
```

```python
# how to give Title and increate font size
import matplotlib.pyplot as plt
Day = [1, 2, 3, 4, 5, 6, 7]
No = [2, 3, 1, 4, 5, 3, 6]

plt.scatter(Day, No,color="r")

plt.title("Scatter Plot", fontsize=20)
plt.xlabel("Day", fontsize=12)
plt.ylabel("No", fontsize=12)
plt.grid(True)
```

```python
# Change the colors of all dots
import matplotlib.pyplot as plt
Day = [1, 2, 3, 4, 5, 6, 7]
No = [2, 3, 1, 4, 5, 3, 6]
colors=["r","y","g","b","r","g","r"]
sizes=[400,200,400,300,200,100,600]
plt.scatter(Day, No,c=colors,s=sizes)

plt.title("Scatter Plot", fontsize=20)
plt.xlabel("Day", fontsize=12)
plt.ylabel("No", fontsize=12)
plt.grid(True)
```

```python
# Change the Transparancy of Graph
import matplotlib.pyplot as plt
Day = [1, 2, 3, 4, 5, 6, 7]
No = [2, 3, 1, 4, 5, 3, 6]
colors=["r","y","g","b","r","g","r"]
sizes=[400,200,400,300,200,100,600]
plt.scatter(Day, No,c=colors,s=sizes,alpha=0.4)

plt.title("Scatter Plot", fontsize=20)
plt.xlabel("Day", fontsize=12)
plt.ylabel("No", fontsize=12)
plt.grid(True)
```

```python
# how to change the Marker
import matplotlib.pyplot as plt
Day = [1, 2, 3, 4, 5, 6, 7]
No = [2, 3, 1, 4, 5, 3, 6]
colors=["r","y","g","b","r","g","r"]
sizes=[400,200,400,300,200,100,600]
plt.scatter(Day, No,c=colors,s=sizes,marker="*",edgecolor="g")

plt.title("Scatter Plot", fontsize=20)
plt.xlabel("Day", fontsize=12)
plt.ylabel("No", fontsize=12)
plt.grid(True)
```

```python
import matplotlib.pyplot as plt
Day = [1, 2, 3, 4, 5, 6, 7]
No = [2, 3, 1, 4, 5, 3, 6]
colors=[10,49,30,29,56,20,30]
sizes=[400,200,400,300,200,100,600]
plt.scatter(Day, No,c=colors,s=sizes,cmap="viridis")
plt.colorbar()
t=set_label=("ColorBar")

plt.title("Scatter Plot", fontsize=20)
plt.xlabel("Day", fontsize=12)
plt.ylabel("No", fontsize=12)
plt.grid(True)
```
