#### Visualization with Matplotlib

```python
# import pyplot library for graph
import matplotlib.pyplot as plt
```

```python
pbs
```

```python
# want to show above table in Bar Chart Graph
plt.bar(pbs['Segment'],pbs['Profit'])
```

```python
plt.figure(figsize=(6,4))
plt.bar(pbs['Segment'],pbs['Profit'])
plt.show()
```

```python
# Modify the Graph
plt.figure(figsize=(7,4))
plt.bar(pbs['Segment'],pbs['Profit'])
plt.title("Profit By Segment",loc='right')
plt.xlabel("Segment",loc='right')
plt.ylabel("Profit")
plt.show()
```

```python
# Color Modification
# Modify the Graph
plt.figure(figsize=(7,4))
plt.bar(pbs['Segment'],pbs['Profit'],color=['green','red','green','green','green'])
plt.title("Profit By Segment",loc='right')
plt.xlabel("Segment",loc='right')
plt.ylabel("Profit")
plt.show()
```

```python
colors=pbs['Profit'].map(lambda val:'lightgreen' if val>0 else 'red')
```

```python
# Color Modification
# Modify the Graph
plt.figure(figsize=(7,4))
plt.bar(pbs['Segment'],pbs['Profit'],color=colors)
plt.title("Profit By Segment",loc='right')
plt.xlabel("Segment",loc='right')
plt.ylabel("Profit")
plt.show()
```

```python
# Date Wise Profit
pbd=df.groupby("Date")['Profit'].sum().reset_index()
```

```python
pbd
```

```python
# How we can create Line Chart
plt.figure(figsize=(12,3))
plt.plot(pbd['Date'],pbd['Profit'])
plt.title("Profit by Date/Month")
plt.xlabel("Date/Month")
plt.ylabel("Profit in $USD")
plt.show()
```

```python
# if we want Dots instead of line
plt.figure(figsize=(12,3))
plt.plot(pbd['Date'],pbd['Profit'],'o')
plt.title("Profit by Date/Month")
plt.xlabel("Date/Month")
plt.ylabel("Profit in $USD")
plt.show()
```

```python
# if we see the graph in start
plt.figure(figsize=(12,3))
plt.plot(pbd['Date'],pbd['Profit'],'*')
plt.title("Profit by Date/Month")
plt.xlabel("Date/Month")
plt.ylabel("Profit in $USD")
plt.show()
```

```python
plt.figure(figsize=(12,3))
plt.plot(pbd['Date'],pbd['Profit'],'s')
plt.title("Profit by Date/Month")
plt.xlabel("Date/Month")
plt.ylabel("Profit in $USD")
plt.show()
```

```python
plt.figure(figsize=(12,3))
plt.plot(pbd['Date'],pbd['Profit'],'.')
plt.title("Profit by Date/Month")
plt.xlabel("Date/Month")
plt.ylabel("Profit in $USD")
plt.show()
```

```python
# Add Markar with Line
plt.figure(figsize=(12,3))
plt.plot(pbd['Date'],pbd['Profit'],marker='o',mfc='red',ms=10)
plt.title("Profit by Date/Month")
plt.xlabel("Date/Month")
plt.ylabel("Profit in $USD")
plt.show()
```

```python
# Change Markar edges color
# Add Markar with Line
plt.figure(figsize=(12,3))
plt.plot(pbd['Date'],pbd['Profit'],marker='o',mec='black',mfc='red',ms=10,color='blue')
plt.title("Profit by Date/Month")
plt.xlabel("Date/Month")
plt.ylabel("Profit in $USD")
plt.show()
```

```python
# How to change Line Style
plt.figure(figsize=(12,3))
plt.plot(pbd['Date'],pbd['Profit'],marker='o', linestyle='dotted',mec='black',mfc='red',ms=10,color='blue')
plt.title("Profit by Date/Month")
plt.xlabel("Date/Month")
plt.ylabel("Profit in $USD")
plt.show()
```
